theme_madhatter \<- function(base_size = 10, exhibit = TRUE) { th \<-
ggplot2::theme_bw(base_size = base_size) + ggplot2::theme( plot.margin =
ggplot2::margin(t = 1, r = 2, b = 1, l = 1, unit = “cm”), axis.title =
ggplot2::element_text( family = “Roboto Condensed”, size = 10 ),
axis.text = ggplot2::element_text( family = “Roboto Condensed”, size = 8
), legend.text = ggplot2::element_text( family = “Roboto Condensed”,
size = 6 ), legend.title = ggplot2::element_text( family = “Alegreya”,
size = 8 ), plot.title = ggplot2::element_text( family = “Alegreya”,
size = 14, face = “bold” ), plot.title.position = “plot”, plot.subtitle
= ggplot2::element_text( family = “Alegreya”, size = 10 ), plot.caption
= ggplot2::element_text( family = “Alegreya”, size = 6 ),
plot.caption.position = “plot” )

if (isTRUE(exhibit)) { th \<- th + ggplot2::theme( plot.background =
ggplot2::element_rect( fill = “bisque”, colour = “black”, linewidth = 1
) ) }

th }
